Arithmetic operators on arrays apply elementwise. A new array is created and filled with the result.

```python
# example 1

import numpy as np

a = np.array([1, 2, 3, 4])
b = np.array([5, 6, 7, 8])

print('a + b =', a+b)
print('a - b =', a-b)
print('a * b =', a*b)
print('a / b =', a/b)
print('a ** 2 =', a**2)
```

```python
# example 2

a = np.array([20, 30, 40, 50])
b = np.arange(4)
c = a - b

print('c =', c)
print('a ** 2 =', a**20)
print('a < 35 =', a<35)
```

Matrix multiplication is <u>**not**</u> the same as elementwise multiplication. The matrix product can be perfomed using the `@`, `np.dot()` or `.dot()` operator.  

<br>

<p style="text-align: center; font-style: italic;">
important rules !
</p>

- For `a @ b`, the number of columns in a must equal number of rows in b

- If a is (mxn) and b is (nxp), result will be (mxp)

```python
# example 3

import numpy as np

u = np.array([[1, 2], [3, 4]])
v = np.array([[5, 6], [7, 8]])

print('u @ v:')
print(u@v)

print('\nnp.dot:')
print(np.dot(u,v))

print('\n.dot:')
print(u.dot(v))
```

```python
# the calculous !!

#  u = [1, 2], [3, 4]
# v = [5, 6], [7, 8]

# Result[0,0] = (1×5) + (2×7) = 5 + 14 = 19
# Result[0,1] = (1×6) + (2×8) = 6 + 16 = 22
#Result[1,0] = (3×5) + (4×7) = 15 + 28 = 43
# Result[1,1] = (3×6) + (4×8) = 18 + 32 = 50
```

```python
# example 4 - compare w/ elementwise multiplication

import numpy as np

u = np.array([[1, 2], [3, 4]])
v = np.array([[5, 6], [7, 8]])

print('u * v:')
print(u*v)

print('\nu @ v:')
print(u@v)
```
