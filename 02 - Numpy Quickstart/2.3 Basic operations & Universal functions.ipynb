Arithmetic operators on arrays apply elementwise. A new array is created and filled with the result.

```python
# example 1

import numpy as np

a = np.array([1, 2, 3, 4])
b = np.array([5, 6, 7, 8])

print('a + b =', a+b)
print('a - b =', a-b)
print('a * b =', a*b)
print('a / b =', a/b)
print('a ** 2 =', a**2)
```

```python
# example 2

import numpy as np

a = np.array([20, 30, 40, 50])
b = np.arange(4)
c = a - b

print('c =', c)
print('a ** 2 =', a**20)
print('a < 35 =', a<35)
```

Matrix multiplication is <u>**not**</u> the same as elementwise multiplication. The matrix product can be perfomed using the `@`, `np.dot()` or `.dot()` operator.  

<br>

<p style="text-align: center; font-style: italic;">
important rules !
</p>

- For `a @ b`, the number of columns in a must equal number of rows in b

- If a is (mxn) and b is (nxp), result will be (mxp)

```python
# example 3

import numpy as np

u = np.array([[1, 2], [3, 4]])
v = np.array([[5, 6], [7, 8]])

print('u @ v:')
print(u@v)

print('\nnp.dot:')
print(np.dot(u,v))

print('\n.dot:')
print(u.dot(v))
```

```python
# the calculous !!

#  u = [1, 2], [3, 4]
# v = [5, 6], [7, 8]

# Result[0,0] = (1×5) + (2×7) = 5 + 14 = 19
# Result[0,1] = (1×6) + (2×8) = 6 + 16 = 22
#Result[1,0] = (3×5) + (4×7) = 15 + 28 = 43
# Result[1,1] = (3×6) + (4×8) = 18 + 32 = 50
```

```python
# example 4 - compare w/ elementwise multiplication

import numpy as np

u = np.array([[1, 2], [3, 4]])
v = np.array([[5, 6], [7, 8]])

print('u * v:')
print(u*v)

print('\nu @ v:')
print(u@v)
```

Some operations act in place to modify an existing array rather than create a new one.

- `+=` Add and assign
- `-=` Subtract and assign
- `*=` Multiply and assign
- `//=` Divide and assign

In-place operations changes the original array, are faster, use less memory (but lose the original) and are great for performance with large data.

Regular operations creates a new array, you keep the original data.

```python
# example 5

import numpy as np

f = np.array([1, 2, 3, 4])

f +=2
print('+=2:', f)

f *=3
print('*=3:', f)

f -=1
print('-=1:', f)

f //=2
print('//=2:', f)
```

```python
# example 6 - WATCH OUT !!!

import numpy as np 

k = np.array([1, 2, 3])
b = k 

b +=10 #this modify BOTH arrays !!!
print('k:', k) #changed
print('b:', b)
```

```python
# example 7 - solution !! 

# use .copy()

import numpy as np 

k = np.array([1, 2, 3])
b = k.copy() #creates copy

b +=10
print('k:', k) #safe !!
print('b:', b)
```

```python
# example 8

import numpy as np

rg = np.random.default_rng(1)  # create instance of default random number generator
a = np.ones((2, 3), dtype=int)
b = rg.random ((2, 3))

a *=3
print('a *= :')
print (a)

b += a
print('\nb += a:')
print(b)
```

Many <u>unary operations</u>, such as computing the sum of all the elements in the array, are implemented as methods of the `ndarray` class.

Unary operations = Operations that work on <u>**one**</u> array and give you back a **single result** or **new transformed array.**

- `.sum():` add everything, all numbers togheter
- `.mean():` find the average, add al numbers an divide by how many they are
- `.min():` find the smallest number
- `.max():` find the biggest number
- `.std():` are numbers close togheter or far apart? how spread out are the numbers? (std ≈ 1-2 = very similar ; std ≈ 10-15 = quite different ; std ≈ 20+ = very different)

```python
# example 9

import numpy as np

h = np.array([150, 200, 175, 300, 250])

total = h.sum()
print('sum:', total) #150+200+175+300+250

av_h = h.mean()
print('avarege:', av_h) #(150+200+175+300+250)/5

min_h = h.min()
print('min:', min_h)

max_h = h.max()
print('max:', max_h)

print('---')

a = np.array([85, 87, 86, 88, 84]) #close togheter
b = np.array([60, 95, 75, 100, 70]) #far apart

print('a spread:', a.std())
print('b spread:', b.std())
```

```python
# example 10

import numpy as np 

m = np.arange(15).reshape(3, 5)
print('m:')
print(m)

print('sum of each column:', m.sum(axis=0))

print('min of each row:', m.min(axis=1))
```

NumPy provides familiar mathematical functions such as sin, cos, and exp. In NumPy, these are called “universal functions” (`ufunc`). Within NumPy, these functions operate elementwise on an array, producing an array as output.
This means thtat universal functions are math functions that work on every element in an array !

- `np.exp():` exponential function
- `np.sqrt():` square root
- `np.add():` elementwise addition
- `np.square():` square
- `np.log():` logarithm
- `np.sin():` sin
- `np.cos():` cos <br> ...

```python
# example 1

import numpy as np

wonderful = np.array([0, 1, 2])
print('wonderful original:', wonderful)

resultexp = np.exp(wonderful)
print('afer np.exp(wonderful):', resultexp) #e^0, e^1, e^2

resultsqrt = np.sqrt(wonderful)
print('after np.sqrt(wonderful):', resultsqrt) #√0, √1, √2

print('---')

m = np.array([0, 1, 2])
n = np.array([2, -1, 4])

print('array m:', m)
print('array n:', n)

resultmn = np.add(m, n)
print('after np.add(m, n):', resultmn) #0+2, 1+(-1), 2+4
```

```python
# example 3

y = np.array([1, 4, 9, 16])

print('original array:', y)
print('square:', np.square(y))
print('logarithm:', np.log(y))

angles = np.array([0, 30, 45, 180])
radians = np.radians(angles) #convert to radians

print('angles:', angles)
print('sin:', np.sin(radians))
print('cos:', np.cos(radians))
```
