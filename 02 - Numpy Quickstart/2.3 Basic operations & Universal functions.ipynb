Arithmetic operators on arrays apply elementwise. A new array is created and filled with the result.

```python
# example 1

import numpy as np

a = np.array([1, 2, 3, 4])
b = np.array([5, 6, 7, 8])

print('a + b =', a+b)
print('a - b =', a-b)
print('a * b =', a*b)
print('a / b =', a/b)
print('a ** 2 =', a**2)
```
