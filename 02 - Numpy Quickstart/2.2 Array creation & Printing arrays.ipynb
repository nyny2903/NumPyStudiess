There are several ways to create arrays. The most commom functionfor this, for example, is `np.array()`

```python
# example 1

import numpy as np 

#creating array
a = np.array([2, 3, 4]) 

#informations abt
print(a)
print (a.dtype)

b = np.array([1.2, 3.5, 5.1])
print(b)
print(b.dtype)

c = np.array([[5, 6, 7], [8, 9, 1]])
print(c)
print(c.dtype)
```

```python
# example 2

a = np.array(1, 2, 3, 4)   #Wrong !!!
```

```python
# example 3

#We can also create a array n specify what we want

n = np.array([[1, 2, 3], [4, 5, 6]], dtype=complex)
print(n)
```

Often, the elements of an array are originally unknown, but its size is known. Hence NumPy offers several functions to create arrays with initial placeholder content (zeros, ones or uninitializad values).

`np.zeros(shape):` Creates an array filled with zeros.<br><br>`np.ones(shape):` Creates an array filled with ones.<br><br>`np.empty(shape):` Creates an array whose initial content is random and depends on the state of the memory.<br><br>`np.full(shape. fill_value):` Creates an array filled with the specified `fill_value`. 

! By default, the dtype of the created array is `float64`, but it can be specified via the key word argument `dtype`.

```python
# example 4

import numpy as np

np.zeros ((3,4))
```

```python
# example 5

import numpy as np

np.ones((2, 3, 4), dtype=np.int16)
```

```python
# example 6

import numpy as np

np.empty ((2, 2))
```

```python
# example 7

import numpy as np

np.full((3, 3), 7)
```

To create sequences of numbers, NumPy provides the `arange` function which is analogous to the Python built-in `range`, but returns an array.
