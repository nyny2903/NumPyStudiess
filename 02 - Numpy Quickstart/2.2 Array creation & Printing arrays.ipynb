There are several ways to create arrays. The most commom functionfor this, for example, is `np.array()`

```python
# example 1

import numpy as np 

#creating array
a = np.array([2, 3, 4]) 

#informations abt
print(a)
print (a.dtype)

b = np.array([1.2, 3.5, 5.1])
print(b)
print(b.dtype)

c = np.array([[5, 6, 7], [8, 9, 1]])
print(c)
print(c.dtype)
```

```python
# example 2

a = np.array(1, 2, 3, 4)   #Wrong !!!
```

```python
# example 3

#We can also create a array n specify what we want

n = np.array([[1, 2, 3], [4, 5, 6]], dtype=complex)
print(n)
```

Often, the elements of an array are originally unknown, but its size is known. Numpy offers several several functions to create arrays with initial content (zeros, ones or uninitializad values)
