**One-dimensional** arrays can be indexed, sliced and iterated over, much like [lists](https://docs.python.org/3/tutorial/introduction.html#lists) and other Python sequences:

```python
# example 1

import numpy as np

a = np.arange(10)**3 #1D array
print('original array:', a)

print('a[2]:', a[2]) #indexing (gets the element at position 2)
print('a[2:5]:', a[2:5]) #slicing (gets a range of elements (position 2 to 5 NOT including 5))

#modify elements
a[:6:2] = -1000 #start to position 6, set every 2nd element to -1000
print('after modification:', a)
```

```python
#example 2

import numpy as np

a = np.arange(10)**3 #1D array
print('original array:', a)

#reverse the array
print('reversed:', a[::-1])

#iteration
print('iterating:')
for element in a:
    print(element)
```

```python
# example 3 - more slicing examples 

import numpy as np

a = np.arange(10)**3 
print('original array:', a)

print('a[1:8:2]:', a[1:8:2]) #[start:stop:step] - 1 to 8, every 2nd element
print('a[:5]:', a[:5]) #from start to position 5
print('a[5:]:', a[5:]) #from position 5 to end
print('a[::2]:', a[::2]) #every 2nd element frrom entire array
```

**Multidimensional** arrays can have one index per axis. These indices are given in a tuple separated by commas.

<p style="text-align: center; font-style: italic;">
important !
</p>

`,` (comma) - separates dmensions/axes

`:` (colon) - indicates slicing within a dimension

```python
# example 4

#create a 2d array
def f(x, y): #for shape (5, 4), x contains row, y contains column
    return 10*x + y 

#for each position, it calculates 10*row_index + column_index:
#row 0: 10*0+0=0, 10*0+1=1, 10*0+2=2, 10*0+3=3
#row 1: 10*1+0=10, 10*1+1=11, 10*1+2=12, 10*1+3=13

b = np.fromfunction(f,(5, 4), dtype=int) 
print('2D array:')
print(b)

#u cant try np.fromfunction(just_rows, (shape), dtype)
# np.fromfunction(just_cols, (shape), dtype)
# np.fromfunction(sum_coord, (shape), dtype)
# np.fromfunction 'says': for each position in my array, calculate the value using this formula based on its row and column position.

print('\nb[2, 3]:', b[2, 3]) #row index = 2 ; column index = 3
print('b[:, 1]:', b[:, 1]) # get all rows, but only column 1
print('b[0:5, 1]:', b[0:5, 1]) #each row in the second column
print('b[:, 1]:', b[:, 1]) #same as above
print('\nb[1:3, :]:')
print(b[1:3, :]) #second and third rows

print('\nb[::2, ::2]:')
print(b[::2, ::2]) #get every 2nd row and every 2nd column
```

```python
# example 5 slicing with ellipsis and new axis

u = np.array([[[0, 1, 2], [10, 12, 13,]], [[100, 101, 102], [110, 112, 113]]])
print('original shape:')
print(u)

print('\nu[1, ...]:')
print(u[1, ...]) #same as u[1, :, :] or u[1]

print('\nu[..., 2]:')
print(u[..., 2]) #same as u[:, :, 2]
```
