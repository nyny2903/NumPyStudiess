`NumPy` = Numerical Python

`Why Numpy?` Because NumPy array is the most efficient way to represent and manipulate a grid of numbers (digital image = grid of numbers) in Python. Instead of worrying about slow loops, you can think in terms of matrices and vectors, which is gihow image processing algorithms are conceptually designed. We cant do efficient image processing in Python wht Numpy. It provides the core data structure (the ndarray) and tools for working with numerical data that are used by nearly every other scientific library, including:
- Pandas
- SciPy
- Scikit-learn
- Matplotlib

`ndarray:`
- Numpy's array class;
- Multidimensional data structure (N-dimensional n>=2 array);
- Allows vectorized operations (wth Python loops, making the code fastes n more conscise);
- Homogeneity ( all elements in the array are of the same data type);
- Efficient way to represent and manipulate numerical data in Python.

`axes` = NumPy dimensions
- 1 axis = a list (vector)
- 2 axes = a table (matrix)
- 3 axes = a cube of data

---

*→ exemple 1* 

[1, 2, 1]

- 3D point w/ 1 axis
- 1 dimensional array, vector
- Lenght of that axis = 3 (has 3 elements inside it) 
- Shape = 3

---

*→ exemple 2*

[ [1.,0.,0.],

[0.,1.,2.] ]

- First axis → axis 0, vertical direction, rows
    - Lenght = 2 (2 rows inside principal list)

<br>

- Second axis → axis 1, horizontal direction, columns
    - Lenght = 3 (3 collumns, internal list has 3 numbers)

<br>

- Shape = (2,3) → 2 rows, 3 columns
